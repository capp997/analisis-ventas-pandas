```python
import pandas as pd

df = pd.read_csv("ventas_sucias.csv")

print(df.shape)          # cuántas filas y columnas hay
print(df.isna().sum())   # dónde faltan datos

df = df.drop_duplicates()  # quita las filas repetidas

df["precio"] = df["precio"].fillna(df["precio"].median())  # rellena precios faltantes con la mediana
df["vendedor"] = df["vendedor"].fillna("Sin asignar")      # rellena vendedores faltantes

df = df.reset_index(drop=True)

print(df)

df["total"] = df["cantidad"] * df["precio"]

print(df.groupby("producto")["total"].sum())

ventas = df.groupby("producto")["total"].sum()

ventas.plot(kind="bar", title="Ventas por producto", ylabel="Total")


```
